```python
# =========================
# Orquestrador Bronze
# =========================

import datetime as dt

print(
    f"🚀 Início da execução Bronze: "
    f"{dt.datetime.now()}"
)

# ==========================================
# Databricks
# ==========================================

try:

    dbutils.notebook.entry_point

    print(
        "✅ Ambiente Databricks detectado."
    )

    # ------------------------------------------
    # Estados IBGE
    # ------------------------------------------

    print(
        "▶ Executando bronze_ibge_estados"
    )

    dbutils.notebook.run(
        "./bronze_ibge_estados",
        timeout_seconds=3600
    )

    # ------------------------------------------
    # Municípios IBGE
    # ------------------------------------------

    print(
        "▶ Executando bronze_ibge_municipios"
    )

    dbutils.notebook.run(
        "./bronze_ibge_municipios",
        timeout_seconds=3600
    )

    # ------------------------------------------
    # INEP Alfabetização
    # ------------------------------------------

    print(
        "▶ Executando bronze_inep_alfabetizacao"
    )

    dbutils.notebook.run(
        "./bronze_inep_alfabetizacao",
        timeout_seconds=3600
    )

# ==========================================
# Execução Local
# ==========================================

except NameError:

    print(
        "⚠️ dbutils não encontrado."
    )

    print(
        "⚠️ O orquestrador só funciona "
        "dentro do Databricks."
    )

    print(
        "⚠️ Execute os notebooks "
        "individualmente no VSCode."
    )

# ==========================================
# Finalização
# ==========================================

print(
    f"✅ Camada Bronze finalizada: "
    f"{dt.datetime.now()}"
)
```
